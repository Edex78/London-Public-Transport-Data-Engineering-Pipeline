```python
from pyspark.sql.functions import col
```

```python
bronze_df = spark.read.table("dbo.public_transport_journeys")
```

```python
silver_df = bronze_df.select(
    "Reporting_Period",
    "Period_and_Financial_year",
    "Bus_journeys_m",
    "Underground_journeys_m",
    "DLR_Journeys_m",
    "Tram_Journeys_m",
    "Overground_Journeys_m",
    "London_Cable_Car_Journeys_m",
    "TfL_Rail_Journeys_m"
)
```

```python
silver_df.write.format("delta").mode("overwrite").saveAsTable("public_transport_silver")
```

```python
silver_df = silver_df.dropDuplicates()
```

```python
from pyspark.sql.functions import sum

silver_df.select(
    *[
        sum(col(c).isNull().cast("int")).alias(c)
        for c in silver_df.columns
    ]
).show()
```

```python
silver_df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("public_transport_silver")
```

```python
from pyspark.sql.functions import (
    col,
    coalesce,
    lit
)
```

```python
from pyspark.sql.functions import col, coalesce, lit

gold_df = silver_df.withColumn(
    "Total_Transport_Journeys",
    coalesce(col("Bus_journeys_m"), lit(0)) +
    coalesce(col("Underground_journeys_m"), lit(0)) +
    coalesce(col("DLR_journeys_m"), lit(0)) +
    coalesce(col("Tram_journeys_m"), lit(0)) +
    coalesce(col("Overground_journeys_m"), lit(0)) +
    coalesce(col("London_Cable_Car_Journeys_m"), lit(0)) +
    coalesce(col("TfL_Rail_Journeys_m"), lit(0))
)

display(gold_df)
```

```python
gold_df.write.format("delta").mode("overwrite").saveAsTable("public_transport_gold")
```
